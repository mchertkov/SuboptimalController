## Graph-local truncation

The theory predicts a squared exponential loss when the relevant action sensitivity decays exponentially.  On case39 the selected H3 controller is weak, so truncation errors are already only $10^{-5}$--$10^{-4}$ and are not cleanly monotone at very small radii.  We therefore interpret this as evidence that locality is inexpensive in this example, **not** as a measurement of the theoretical decay exponent.  Case118 is the more meaningful scaling test.

## Current certification status

The case39 campaign is complete as an **experiment**: noise sweeps, controller hierarchy, smooth PIC candidate, curvature diagnostics, locality, implementation defects, numerical convergence, and terminal/tube bounds have all been produced.
