## Actuator hierarchy

The four actuator subspaces are: H0 one physical slack generator; H1 a distributed rank-one mode; H2 three generator modes including low-order modes of the Kron-reduced generator stiffness; H3 independent generator control.  The mode columns are orthonormal so the same scalar control price means the same physical generator-space quadratic norm.

```python
h=pd.read_csv(RES/'h0_h3.csv'); h
```

## PIC matching and candidate gaps

For a generator mode matrix $A$, the maximal linearly solvable diffusion satisfies
$$\Gamma^2-\frac{\lambda}{r}S_gAA^TS_g^T\succeq0.$$
The PIC value is a rigorous lower bound only after the deleted-noise curvature condition is verified on the relevant domain.  The table therefore labels $J-W$ as a **candidate** certificate gap.

## PIC feedback projection

An independent sparse PIC feedback estimate was projected onto the full H3 LQR command.  After excluding the ill-conditioned $t=0$ point, the projection is near $10^{-2}$ for both noise models.  This supports the weak H3 LQR scale selected by the direct SOC scan.

## Smooth versus stopped formulation

Near the cohesion boundary, an absorbing failure cost turns Feynman--Kac estimation into a rare-survival problem.  The smooth extended barrier keeps the performance PDE statistically regular even when trajectories cross $90^\circ$; safety is then handled separately.  The paper therefore uses the smooth formulation for PIC performance.

```python
r=json.loads((RES/'radial_comparison.json').read_text()); pd.DataFrame(r)
```

## Deleted-noise curvature screen

For case39 the nonlinear residual reduces to
$$\rho=\operatorname{tr}[(D-\widetilde D)_{\omega\omega} W_{\omega\omega}].$$
The current representative compact-domain screen is positive for every tested state and every tested Hutchinson direction.  This is strong numerical evidence but not yet a domain-wide theorem.

```python
c=json.loads((RES/'curvature_screen.json').read_text()); print('LCB minima',c['homogeneous_lcb_min'],c['activity_lcb_min'])
```

## Analytic terminal layer and local sensitivity

The terminal curvature is analytic.  A cycle-consistent, energy-coupled enclosure gives a rigorous positive-curvature terminal layer.  In the interior, first/second/third flow variations are propagated in local mechanical metrics; this avoids the catastrophic conservatism of one global $84^\circ$ sector norm.
